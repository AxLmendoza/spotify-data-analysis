```python
#Cuales son los 10 generos con mas canciones?
df_spoty.value_counts("track_genre").head(10)
```

```python
#Los 10 artistas con mas canciones
df_spoty["artists"].value_counts().head(10)
```

```python
#Las 10 canciones más populares.
df_spoty[["popularity", "artists", "track_name"]].sort_values(by="popularity", ascending=False).head(10)
```

```python
df_spoty["artists"].value_counts().head(10)
```

```python
df_spoty
```

```python
# ¿Cuál es la popularidad promedio POR GÉNERO?
df_spoty.groupby("track_genre")["popularity"].mean()
```

```python
df_spoty
```

```python
#¿Cuál es el género con MAYOR popularidad promedio?
df_spoty.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(1)
```

```python
#¿Cuál es el género con MENOR popularidad promedio?
df_spoty.groupby("track_genre")["popularity"].mean().sort_values(ascending=True).head(1)
```

```python
#¿Cuál es la duración promedio (duration_ms) por género?
df_spoty.groupby("track_genre")["duration_ms"].mean()
```
